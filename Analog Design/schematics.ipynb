```python
import schemdraw
import schemdraw.elements as elm
```

## Lesson 1

### Common Source

```python
with schemdraw.Drawing() as d:
    # Components
    r = d.add(elm.ResistorIEC(unit=1).theta(90).label('$R_{L}$', loc='bottom'))
    d.move(0, -2)
    m1 = d.add(elm.NFet().theta(0).reverse().label('$M_{1}$', loc='right'))
    gnd = d.add(elm.GroundSignal().at(m1.source))
    vdd = d.add(elm.Vdd().at(r.end).label('$V_{DD}$', loc='top'))

    # Connections
    d.add(elm.Wire().at(m1.drain).to(r.start))
    d.add(elm.Wire().at((m1.drain[0],m1.drain[1]+0.4)).to((m1.drain[0]+0.5,m1.drain[1]+0.4)).label('$V_{OUT}$', loc='right'))
    d.add(elm.Wire().at((m1.gate[0]-0.1,m1.gate[1])).to(m1.gate).label('$V_{IN}$', loc='left'))
```
